```python
nav_response = {
    "fund_id": "FG-101",
    "reported_nav": 100000,
    "holdings": [
        {"symbol": "AAPL", "quantity": 200, "price": 170},
        {"symbol": "TSLA", "quantity": 50, "price": 800},
        {"symbol": "MSFT", "quantity": 100, "price": 310}
    ]
}
```

```python
nav = nav_response['holdings']

nav_df = pd.DataFrame(nav)

print(nav_df)
```

```python
nav_df['nav'] = nav_df['quantity'] * nav_df['price']

print(nav_df)

total_nav = nav_df['nav'].sum()
print('Total NAV for fund is:', total_nav)

```

```python
reported_nav = nav_response['reported_nav']

if total_nav == reported_nav:
    print('NAV is correct')
else:
    print('NAV is incorrect')
    print('Reported NAV:', reported_nav)
    print('Calculated NAV:', total_nav)
    print('Difference:', reported_nav - total_nav)
```

```python
nav_response = {
    "fund_id": "FG-101",
    "reported_nav": 100000,
    "holdings": [
        {"symbol": "AAPL", "quantity": 200, "price": 170},
        {"symbol": "TSLA", "quantity": 50, "price": 800},
        {"symbol": "MSFT", "quantity": 100, "price": 310}
    ]
}

holdings = nav_response['holdings']
hold = pd.DataFrame(holdings)
print(hold)
```

```python
hold['position_value'] = hold['quantity'] * hold['price']
total_position_nav = hold['position_value'].sum()
print(hold)
print(total_position_nav)
```

```python
if total_position_nav == nav_response['reported_nav']:
    print('NAV is correct')
else:
    print('NAV is incorrect')
    print('Reported NAV:', reported_nav)
    print('Calculated NAV:', total_nav)
    print('Difference:', reported_nav - total_nav)
```
