```python
import pandas as pd
import numpy as np


new_nav_data = {
    "fund_id": "FG-202",
    "reported_nav": 220000,
    "holdings": [
        {"symbol": "GOOGL", "quantity": 50, "price": 2900},
        {"symbol": "AMZN", "quantity": 30, "price": 3300},
        {"symbol": "META", "quantity": 100, "price": 310}
    ]
}

data = new_nav_data["holdings"]
holdings = pd.DataFrame(data)


holdings['total_value'] = holdings['quantity'] * holdings['price']
print(holdings)
total_nav = holdings['total_value'].sum()
reported_nav = new_nav_data["reported_nav"]

print(f'Total NAV: {total_nav}')
print(f'Reported NAV: {reported_nav}')

if total_nav == reported_nav:
    print('NAV is correct')
else:
    print('NAV is incorrect')
    print(f'Difference: {total_nav - reported_nav}')
    print(f'Percentage difference: {((total_nav - reported_nav) / reported_nav) * 100}%')


```
