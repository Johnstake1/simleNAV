```python
import pandas as pd
import numpy as np

trade_json = {
    "account_id": "FG-304",
    "trades": [
        {"symbol": "AAPL", "quantity": 50, "price": 175, "date": "2025-03-28"},
        {"symbol": "MSFT", "quantity": 30, "price": 310, "date": "2025-03-28"},
        {"symbol": "TSLA", "quantity": 20, "price": 800, "date": "2025-03-29"},
        {"symbol": "AAPL", "quantity": 40, "price": 180, "date": "2025-03-29"}
    ]
}

trades = trade_json['trades']
df = pd.DataFrame(trades)

print(df)
```

```python
df['trade_value'] = df['quantity'] * df['price']

group_by_symbol = df.groupby('symbol')['trade_value'].sum()

print(group_by_symbol)
```

```python
group_by_date = df.groupby('date')['trade_value'].sum()

print(group_by_date)
```

```python
df['high_value'] = df['trade_value'].apply(lambda x: 'TRUE' if x > 10000 else 'FALSE')
print(df)
```

```python
import json

save = df.to_json(orient='records', indent=4, lines=True)
print(save)
```
